```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from src.utils.carga import cargar_csv

df=cargar_csv("../data/original/Student Depression Dataset.csv")

df.head()
df.info()
df.describe()
```

```python
valores_n=df.isnull().sum()
valores_n
```

```python
sns.countplot(x="Depression", data=df)
plt.savefig("../outputs/graficos/depresion.png")
plt.show()
```

```python
sns.boxplot(x="Depression", y="Academic Pressure", data=df)
plt.savefig("../outputs/graficos/presion_academica.png")
plt.show()
```

```python
sns.boxplot(x="Depression", y="Work/Study Hours", data=df)
plt.savefig("../outputs/graficos/horas_estudio.png")
plt.show()
```

```python
sns.countplot(x="Sleep Duration", hue="Depression", data=df)
plt.xticks(rotation=45)
plt.savefig("../outputs/graficos/sueno.png")
plt.show()
```

```python
sns.heatmap(df.corr(numeric_only=True), annot=False, cmap="coolwarm")
plt.title("Matriz de correlación")
plt.savefig("../outputs/graficos/correlacion.png")
plt.show()
```
